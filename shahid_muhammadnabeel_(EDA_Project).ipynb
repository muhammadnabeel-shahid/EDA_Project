**Phase 2: Data Cleaning and Preparation**

```python
# Drop irrelevant columns (e.g., page_url or location_id if not used) to streamline the dataset.

property_data = property_data.drop("page_url" , errors="ignore")
```

```python
# Categorical/String columns: Fill Null/NA values with the string "Unknown".
categorical_cols = property_data.select_dtypes(include=['object']).columns
property_data[categorical_cols] = property_data[categorical_cols].fillna("Unknown")
```

```python
# Categorical/String columns: Fill Null/NA values with the string "Unknown".

numerical_cols = property_data.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_cols:
    property_data[col] = property_data[col].fillna(property_data[col].mean())
```

```python
# Verification: Check programmatically if all missing values have been successfully imputed.

property_data.isnull().sum()
```

**Data Transformation**

```python
#

def convert_to_marla(area):
    if isinstance(area, str):
        parts = area.split()
        # Remove commas from the numeric part before converting to float
        value_str = parts[0].replace(',', '')
        value = float(value_str)
        unit = parts[1].lower()

        if unit == "marla":
            return value
        elif unit == "kanal":
            return value * 20

    return None
```

```python
# Create a new Column named Area_in_Marlas

property_data['area_in_marla'] = property_data['area'].apply(convert_to_marla)
```

## Define 'High-Value' Property

### Subtask:
Calculate a suitable price threshold (e.g., the 75th percentile of property prices) to define what constitutes a 'high-value' property. This will create the `high_value` DataFrame.


**Reasoning**:
To define 'high-value' properties, I need to calculate the 75th percentile of the 'price' column as the threshold and then filter the original DataFrame to create the 'high_value' DataFrame.



```python
price_threshold = property_data['price'].quantile(0.75)
high_value = property_data[property_data['price'] >= price_threshold]

print(f"The 75th percentile price threshold is: {price_threshold}")
print(f"Number of high-value properties: {len(high_value)}")
print("First 5 rows of high_value properties:")
high_value.head()
```

## Identify Active Agents in High-Value Market

### Subtask:
Using the newly defined `high_value` DataFrame, identify the top 10 agents with the most listings within this segment. This will recreate the `high_agents` variable.


**Reasoning**:
To identify the top 10 agents with the most listings in the high-value segment, I will count the occurrences of each agent in the 'agent' column of the `high_value` DataFrame and then select the top 10 results.



```python
high_agents = high_value["agent"].value_counts().head(10)

print("Top 10 Agents in High-Value Market:")
print(high_agents)
```
